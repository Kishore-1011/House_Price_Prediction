# 2. Predicting a client's file

Loads the saved model, checks a client's CSV, predicts a price for every house and writes the result.

The saved model was trained on made-up data, so these prices aren't real valuations.

```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))
```

```python
import pandas as pd

from house_price.config import CLIENT_DATA, PREDICTIONS
from house_price.predict import predict_prices
from house_price.storage import load_model
```

The model file's checksum is verified before it's loaded, because loading a joblib file can run code.

```python
model = load_model()
print("Model loaded:", type(model).__name__)
```

## The client's file

```python
client = pd.read_csv(CLIENT_DATA)
client
```

## Predict and save

Missing bedrooms or ages are filled in automatically. Unusable rows stop with a message naming the problem.

```python
result = predict_prices(client, model=model)
result.to_csv(PREDICTIONS, index=False)
result
```
